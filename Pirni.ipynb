```python
df = pd.read_csv('Dataset_Kinerja_Karyawan.csv')
df.info()
```

```python
df.head()
```

```python
print("==Statistika Deskriptif Skor Kinerja ===")
desc_stats = df['Skor_Kinerja'].describe()
print(desc_stats)
```

```python
modus = df['Skor_Kinerja'].mode() [0]
varians = df['Skor_Kinerja'].var()
range_skor = df['Skor_Kinerja'].max() - df['Skor_Kinerja'].min()

print(f"\nmodus: {modus}")
print(f"Varians: {varians}")
print(f"Range: {range_skor}")
```

```python
plt.figure(figsize=(10, 6))
sns.histplot(df['Skor_Kinerja'], bins=20, kde=True, color='skyblue', edgecolor='red')
plt.axvline(df['Skor_Kinerja'].mean(), color='black', linestyle='dashed', linewidth=2, label=f"Mean: {df['Skor_Kinerja'].mean():.2f}")
plt.axvline(df['Skor_Kinerja'].median(), color='green', linestyle='dashed', linewidth=2, label=f"Median: {df['Skor_Kinerja'].median():.2f}")
plt.title('Distribusi Skor KinerjaKaryawan', fontsize=14, pad=15)
plt.xlabel('Skor Kinerja', fontsize=12)
plt.ylabel('Frekuensi', fontsize=12)
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.boxplot(x='Departemen', y='Skor_Kinerja', data=df,palette='Set2')
plt.title('Sebaran Skor Kinerja per Departemen', fontsize=14, pad=15)
plt.xlabel('Departemen', fontsize=12)
plt.ylabel('Skor Kinerja',  fontsize=12)
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.barplot(x='Departemen', y='Lama_Bekerja_Tahun', data=df, errorbar=None, palette='viridis')
plt.title('Rata-rata Lama Bekerja Karyawan per Departemen', fontsize=14, pad=15)
plt.xlabel('Departemen', fontsize=12)
plt.ylabel('Lama Bekerja (Tahun)',  fontsize=12)
plt.show()
```

1. **Sebagian Besar Kinerja Karyawan Terletak di Kisaran Menengah**
Dari 1. 000 karyawan yang disample, rata-rata skor kinerja tercatat sebesar 68,36. Di samping itu, nilai median tercatat pada angka 69,00, dan skor yang paling umum muncul (modus) berada di angka 70. Ini mencerminkan bahwa, secara umum, performa karyawan terpusat pada kisaran 68 hingga 70, yang menunjukkan tingkat kinerja rata-rata yang cukup konsisten dengan peluang untuk perbaikan lebih lanjut melalui program pelatihan atau insentif.
